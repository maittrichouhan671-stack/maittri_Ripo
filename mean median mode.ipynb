```python
import pandas as pd 

#create dataset
data= {
    "Student_Name" : [
        'Aarav', 'Priya','Rahul','Sneha','Arjun','Ananya','Vikram','Meera','Rohan','Ishita'
    ],
    'Marks': [85,92,76,88,67,95,73,84,58,90]

}


#create Dataframe
df=pd.DataFrame(data)

#display dataset
display(df)
```

```python
#calculate mean of marks
mean_marks = df['Marks'].mean()

print("Mean of student marks:",mean_marks)

df
```

```python
# calculate mode of marks
mode_marks = df['Marks'].mode()

print("Mode of student marks:", mode_marks.tolist())

```

```python
median_marks = df['Marks'].median()

print("Median Marks:", median_marks)
```
